## Напишите функцию, которая по строке возвращает целое число — сумму кодов всех символов (коды символов получите с помощью функции ord().
- Проверьте работу функции на разных строках, например:  'cat', 'dog', 'god', 'thing', 'night'
- Что произойдёт для строк, содержащих одинаковые буквы?  

```python
def sum_ord(s):
    total = 0
    for ch in s:
        total += ord(ch)
    return total

for word in ['cat', 'dog', 'god', 'thing', 'night']:
    print(word, sum_ord(word))

print("Одинаковые буквы ('aaa'):", sum_ord('aaa'))
```

## Напишите функцию, которая создаёт словарь (или хеш-таблицу), где:
	•	ключ = слово,
    
	•	значение = сколько раз это слово встретилось. 
## Для короткого текста, предоставленного в ячейке ниже, найдите:
	
    •	самое частое слово,
	
    •	все слова, которые встречаются больше одного раза.

```python
str1 = 'I felt happy because I saw the others were happy and because I knew I should feel happy, but I wasn’t really happy.'
```

```python
def count_words(text):
    clean = text.lower().replace(",", "").replace(".", "").replace("’", "'")
    words = clean.split()
    
    table = {}
    for w in words:
        if w in table:
            table[w] += 1
        else:
            table[w] = 1
    return table


word_counts = count_words(str1)

most_freq_word = max(word_counts, key=word_counts.get)
print("Самое частое слово:", most_freq_word, "—", word_counts[most_freq_word])

more_than_one = [w for w, c in word_counts.items() if c > 1]
print("Встречаются больше одного раза:", more_than_one)
```

## В ячейке ниже дана строка ДНК. Постройте хеш-таблицу(словарь), для всех подстрок длиной 4(подстрока - непрерывный участок строки), в которой ключом будет являться сам фрагмент, а значением - список позиций в ДНК, на которых он встречается. Далее найдите все позиции подстроки 'ATCA'.

```python
dna = "ATCAGGCATCAGGCACATGATCGTACTACATGAATCAGTGAATCGATCAGTACACGT"
```

```python

dna_hash = {}
for i in range(len(dna) - 3):
    frag = dna[i:i + 4]
    if frag not in dna_hash:
        dna_hash[frag] = []
    dna_hash[frag].append(i)

print("позиции 'ATCA':", dna_hash.get("ATCA", []))
```
