## Создайте класс SmartHouse (умный дом), отразив при этом основные принципы ООП. Поясните логику своего кода. 

```python
class Device:
    def __init__(self, name):
        self.name = name
        self.on = False

    def turn_on(self):
        self.on = True
        print(f"{self.name} вкл")

    def turn_off(self):
        self.on = False
        print(f"{self.name} выкл")


class SmartHouse:
    def __init__(self, name):
        self.name = name
        self.devs = []

    def add(self, d):
        self.devs.append(d)
        print(f"Добавлен: {d.name}")

    def all_on(self):
        for d in self.devs:
            d.turn_on()

    def all_off(self):
        for d in self.devs:
            d.turn_off()

    def info(self):
        print(f"Дом: {self.name}")
        for d in self.devs:
            s = "вкл" if d.on else "выкл"
            print(f"  {d.name}: {s}")
```

```python
h = SmartHouse("Мой дом")
l = Device("Лампа")
t = Device("Термостат")
c = Device("Камера")

h.add(l), h.add(t), h.add(c)

h.all_on()
h.info()
h.all_off()
```

Инкапсуляция — данные устройства хранятся внутри объекта;
Наследование — можно создать потомков Device (лампа, термостат и т.д.);
Полиморфизм — all_on() работает с любыми устройствами одинаково;
Абстракция  — дом содержит список устройств;

## Имеется нижепредставленная последовательность. Напишите функцию, которая удаляет лишние символы, убирает пробелы, приводит к верхнему регистру и разбивает строку на триплеты. Результатом функции должна быть строка

```python
raw_seq =  '''        1 ctgactcaaa cgtctcactg atcgttacac aacaccagtg taatttcgca gtgcttatta
       61 acaagctaaa gcgctagcat ttactggaag cacgacttcc tctcgccaac gtggatataa
      121 agctgtatta atcatgggtg ttcacgactt gtggtccatc gtggagccag tccgagaggc
      181 ggtgccgttg tacagtttga gcggaaagac tctggcggta gacctgagct tatgggtctg
      241 tgaggctcag cAcgtccagg ccatgatggg cagagttatg aagccacatc tgaggaattt
      301 atttttccga gtatcatcgc tcacactgat GGGGtgaag ctcgtctttg tcatggaggg
      361 agaagctccc caacttaaat cagaaacaat gagcaaaagg acagaaatga gattcggagg
      421 atccagaaaa gCCcatttct caaagtcatc cacaactacc agcagagggc gctttaaagc
      481 tgtacttcga gGGGtgtgcag atatgttgaa ctacctgggt gtgccctggg tgacagcagc
      541 tggggaagca gaggccatgt gtgcttacct ggactcagca gggctggttg atggctgcat
      601 caccaacgat ggagatgcat tcttatatgg ggctaggacg gtctacagga acttcaatat
      661 gaacagtaaa gaccctcagg tcgattgcta ccagacatcc cgagtgcaga ccgagctaca
      721 cctgtcaaga gagaatcttg ttggcttggc cattctcctc ggttgtgatt aagcagatgt
      781 gttgtataaa gtaat'''



def process_seq(raw_seq):

    cleaned = ""
    for line in raw_seq.split('\n'):
        line = line.lstrip('0123456789 ')
        cleaned += line
    
    cleaned = cleaned.upper().replace(" ", "")
    triplets = [cleaned[i:i+3] for i in range(0, len(cleaned), 3)]
    
    return ' '.join(triplets)
```

```python
# Использование
result = process_seq(raw_seq)
print(result)
```

## Через Biopython реализуйте функцию, которая принимает строку с матричной РНК (из прошлого задания 2.3) и преобразует её в список аминокислот, соответствующих каждому триплету. При этом функция должна начинать считывание с первого встреченного старт-кодона и завершать работу при обнаружении стоп-кодона

```python
from Bio.Seq import Seq

def translate_mrna(mrna_seq):
    start_idx = mrna_seq.find('AUG')
    if start_idx == -1:
        return "Старт-кодон не найден"
    coding_seq = mrna_seq[start_idx:]
    rna = Seq(coding_seq)
    protein = rna.translate(to_stop=True)  # to_stop=True останавливает на стоп-кодоне
    
    return list(protein)
```

## Создайте класс для работы с белковыми последовательностями. В классе должны быть определены три атрибута: 
1) переменная класса, которая хранит количество созданных объектов (белков), 
2) переменная класса в виде словаря, где ключом является идентификатор белка, а значением его аминокислотная последовательность, и  
3) переменная класса с контрольной аминокислотной последовательностью "SEGAFGRDFHNRESHJRSTTTTHHREHSEHEHJ", которая используется для выравнивания. 

В конструкторе класса необходимо инициализировать переменную экземпляра, принимающую строку с аминокислотной последовательностью, передаваемую при создании объекта. Класс должен содержать статический метод, возвращающий словарь с белками и их последовательностями, а также метод, который сравнивает последовательность экземпляра с контрольной строкой и вычисляет процент гомологичности. При этом, если последовательность экземпляра длиннее контрольной, её нужно обрезать до длины контрольной последовательности, а если она короче — дополнить символом "*" до соответствующей длины.

```python
# Для получения рандомных id можно использовать следующий код
import random

class Protein:
    count = 0                        
    id_prot = {}                       
    control = "SEGAFGRDFHNRESHJRSTTTTNHRENSEHEHJ"

    def __init__(self, seq):
        self.seq = seq.upper()
        self.id = random.randint(1, 100)
        Protein.count += 1
        
        Protein.id_prot[self.id] = self.seq


    def sequence(self):
        s = self.seq
        if len(s) > len(Protein.control):
            s = s[:len(Protein.control)] 
        else:
            s = s + "*" * (len(Protein.control) - len(s))
        return s


    def homology(self):
        s = self.sequence()
        matches = sum(1 for a, b in zip(s, Protein.control) if a == b)
        return matches / len(Protein.control) * 100

    @staticmethod
    def get_all():
        return Protein.id_prot

```
